```python
import numpy as np

a = np.array([1., 2., 1.])
b = np.array([2., 1., 1.])
cosine = a @ b / (np.linalg.norm(a) * np.linalg.norm(b))
print('dot=', a @ b, 'cosine=', cosine)
```

```python
X = np.array([[1., 1.], [2., 2.], [3., 3.]])
y = np.array([2., 4., 6.])
print('shape=', X.shape, 'rank=', np.linalg.matrix_rank(X))
print('least-squares W=', np.linalg.lstsq(X, y, rcond=None)[0])
```

```python
w, learning_rate = 0., 0.1
for step in range(5):
    loss = (w - 3) ** 2
    gradient = 2 * (w - 3)
    print(step, round(w, 4), round(loss, 4), round(gradient, 4))
    w -= learning_rate * gradient
```

```python
logits = np.array([2., 1., -1.])
stable = logits - logits.max()
probability = np.exp(stable) / np.exp(stable).sum()
print(probability, probability.sum())
```

## 설명할 것
1. 두 번째 코드에서 inverse 대신 `lstsq`를 쓴 이유
2. Gradient의 부호와 W의 이동 방향
3. Softmax 전 최댓값을 빼도 확률이 같은 이유
